```python
session = session.getOrCreate()
session.sparkContext.setLogLevel("ERROR")
```

```python
session.sql("SELECT * FROM nessie.silver.passengers").show()
```

```python
df_bronze = session.sql("SELECT * FROM nessie.bronze.trains")
```

```python
df_silver = session.sql("SELECT * FROM nessie.silver.trains")
```

```python
df_bronze_cleaned = (df_bronze.withColumn("sk_id",F.abs(F.xxhash64(F.col("id"),F.col("updated_at"))))
.withColumn("name",F.trim(F.lower(F.col("name"))))
.withColumn("type", F.coalesce(F.trim(F.lower(F.col("type"))), F.lit("unknown")))
.withColumn("capacity", F.coalesce(F.col("capacity"), F.lit(0)))
.withColumn("is_active", F.lit(True).cast("boolean"))
.withColumn("start_date", F.to_timestamp(F.col("updated_at")))
.withColumn("end_date", F.lit(None).cast("timestamp"))
.dropDuplicates(["sk_id"])
).select(
                F.col("sk_id"),
                F.col("id"),
                F.col("name"),
                F.col("type"),
                F.col("capacity"),
                F.col("is_active"),
                F.col("start_date"),
                F.col("end_date")
            )

df_bronze_cleaned.createOrReplaceTempView("df_bronze")
df_bronze_cleaned.show()
```

```python
session.sql(
"""
        MERGE INTO {full_table_name} t
        USING {table_view} s
        ON t.id = s.id AND t.is_active = true

        WHEN NOT MATCHED BY SOURCE THEN 
        THEN UPDATE SET 
            t.is_active = false,
            t.end_date = s.start_date

        WHEN MATCHED AND (
            t.name <> s.name OR 
            t.type <> s.type OR 
            t.capacity <> s.capacity
        )
        THEN UPDATE SET 
            t.is_active = false,
            t.end_date = s.start_date


        WHEN NOT MATCHED THEN 
        INSERT (sk_id, id, name, type, capacity, is_active, start_date, end_date)
        VALUES(s.sk_id, s.id, s.name, s.type, s.capacity, s.is_active, s.start_date, null)
"""
)
```

```python
session.sql("""
WITH cte as (
    select t.sk_id from nessie.silver.trains t WHERE not exists (SELECT 1 from df_bronze s WHERE t.sk_id = s.sk_id)
)

UPDATE nessie.silver.trains t
SET 
    t.is_active = false,
    t.end_date = DATE('2024-01-02')
WHERE EXISTS (SELECT 1 FROM cte WHERE t.sk_id = cte.sk_id)
""").show()
```
